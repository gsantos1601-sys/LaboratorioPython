# Tienda en línea con descuentos

📌 **Sección 7 · Capítulo 89**

## 🎯 Planteamiento
Una tienda online quiere aplicar descuentos a sus clientes según dos condiciones: si son miembros de la tienda y el monto de su compra. El programa debe calcular cuánto paga cada cliente después de aplicar el descuento que le corresponde.

## ⚙️ Qué hace el programa
1. Pide el monto de la compra en RD$.
2. Pregunta si el cliente es miembro de la tienda (si/no) y normaliza la respuesta.
3. Si es miembro y compra RD$1,000 o más, aplica un **10 %** de descuento.
4. Si es miembro y compra menos de RD$1,000, aplica un **5 %** de descuento.
5. Si no es miembro, no aplica descuento.
6. Muestra el monto de la factura, el descuento aplicado y el total a pagar.

## 🧰 Conceptos aplicados
`input()` · `float()` · `lower()` y `strip()` · condicionales `if` / `elif` / `else` · operador lógico `and` · comparación `>=` · cálculo de porcentajes

## 💻 Código

```python
# Tienda en línea con descuentos

print('****TIENDA ONLINE****')

# --- Entrada de datos ---
monto_compra = float(input('Ingrese el monto a pagar RD$: '))

# strip() quita espacios y lower() pasa a minúsculas, para que "SI " y "si" valgan igual
asociado = input('¿Usted es miembro de la tienda si/no?').lower().strip()

descuento = 0.0   # valor inicial: sin descuento

# --- Miembro con compra de RD$1,000 o más: 10 % de descuento ---
if monto_compra >= 1000.00 and asociado == 'si':

    descuento = monto_compra * 0.10          # 10 % del monto
    total_pago = monto_compra - descuento    # monto menos el descuento
    print('Monto de la factura:', monto_compra)
    print('Descuento aplicable: 10%')
    print('Total a pagar:', total_pago)

# --- Miembro con compra menor a RD$1,000: 5 % de descuento ---
elif asociado == 'si':

    descuento = monto_compra * 0.05          # 5 % del monto
    total_pago = monto_compra - descuento
    print('Monto de la factura:', monto_compra)
    print('Descuento aplicable: 5%')
    print('Total a pagar:', total_pago)

# --- No es miembro: paga el monto completo ---
else:
    print('Total a pagar:', monto_compra)
```
