# Cajero automático

## 🎯 Planteamiento
Se necesita simular el funcionamiento de un cajero automático. El usuario debe poder retirar dinero, depositar y consultar su balance, y el menú debe repetirse después de cada operación hasta que decida salir. Además, el sistema debe validar lo que se escribe: si el usuario ingresa letras o un monto menor que 1, muestra un error y vuelve a pedir el dato.

## ⚙️ Qué hace el programa
1. Inicia la cuenta con un saldo de RD$1,000 y muestra el menú de opciones.
2. Pide al usuario que elija una opción. Si escribe algo que no sea un número, muestra un error y la vuelve a pedir.
3. **Retirar:** pide el monto (debe ser un número mayor o igual a 1). Si es válido y hay fondos suficientes, lo resta del saldo; si no, avisa que los fondos son insuficientes.
4. **Depositar:** pide el monto (debe ser un número mayor o igual a 1) y lo suma al saldo.
5. **Consultar balance:** muestra el saldo actual.
6. **Salir:** termina el ciclo y muestra un mensaje de cierre.
7. Si el usuario elige un número de opción que no existe, le avisa y vuelve a mostrar el menú.

## 🧰 Conceptos aplicados
`input()` · `int()` · `isdigit()` para validar entradas · ciclos `while` anidados (validación y menú) · `while` con bandera (`salir`) · `while` con `else` · operador `or` · `if` / `elif` / `else` · acumuladores (`+=` y `-=`) · f-strings con formato `:.2f`

## 💻 Código

```python
# Cajero automático

# --- Datos iniciales ---
saldo = 1000       # saldo inicial en RD$
depositar = 0      # monto a depositar
retirar = 0        # monto a retirar
salir = False      # bandera: cuando pasa a True, el menú termina

# --- Menú de opciones (cadena de varias líneas con ''') ---
print('''   ***MENU****

1. Retirar
2. Depositar
3. Consultar balance
4. Salir''')

# --- El menú se repite hasta que el usuario elija salir ---
while not salir:

  # --- Validar la opción: debe ser un número (no letras ni vacío) ---
  # isdigit() es True solo si el texto tiene únicamente dígitos
  opciones = input('\nSeleccione una opcion: ').strip()
  while not opciones.isdigit():
    print('Error: ingrese un numero valido.')
    opciones = input('\nSeleccione una opcion: ').strip()
  opciones = int(opciones)    # ya validado, se convierte a entero

  if opciones == 1:
    # --- Validar el monto: solo dígitos y mayor o igual a 1 ---
    # Si no es número, "or" no llega a evaluar int() y evita el error
    retirar = input('Digite la cantidad a retirar: ').strip()
    while not retirar.isdigit() or int(retirar) < 1:
      print('Error: ingrese un monto numerico mayor o igual a 1.')
      retirar = input('Digite la cantidad a retirar: ').strip()
    retirar = int(retirar)

    # --- No se puede retirar más de lo que hay en la cuenta ---
    if retirar > saldo:
      print('Fondos insuficientes.')
    else:
      saldo -= retirar                                # resta el retiro al saldo
    print(f'Su balance es: RD${saldo:.2f}\n')         # :.2f muestra 2 decimales
  elif opciones == 2:
    # --- Validar el monto: solo dígitos y mayor o igual a 1 ---
    depositar = input('Digite la cantidad a depositar: ').strip()
    while not depositar.isdigit() or int(depositar) < 1:
      print('Error: ingrese un monto numerico mayor o igual a 1.')
      depositar = input('Digite la cantidad a depositar: ').strip()
    depositar = int(depositar)

    saldo += depositar                                # suma el depósito al saldo
    print(f'Su balance es: RD${saldo:.2f}\n')
  elif opciones == 3:
    print(f'Su balance es: RD${saldo:.2f}\n')         # solo consulta, no cambia el saldo
  elif opciones == 4:
    print('Saliendo del sistema\n')
    salir = True                                      # la bandera termina el ciclo
  else:
    print('Digite una opcion valida.\n')

# El else del while se ejecuta cuando el ciclo termina normalmente
else:
  print('El sistema fue finalizado correctamente.')
```

Ejemplo de ejecución: consulta, depósito, retiro, opción inválida y salida.
