# Promedio de calificaciones

📌 **Sección 9 · Capítulo 134**

## 🎯 Planteamiento
Se necesita un programa que registre las calificaciones de un estudiante y calcule su promedio. El usuario indica cuántas calificaciones va a ingresar, las escribe una por una, y el programa las guarda en una lista, las muestra y calcula el promedio.

## ⚙️ Qué hace el programa
1. Pide la cantidad de calificaciones que se van a agregar.
2. Repite la solicitud una vez por cada calificación, la guarda en una lista y la va sumando.
3. Muestra todas las calificaciones ingresadas, una al lado de la otra.
4. Calcula el promedio dividiendo la suma entre la cantidad de calificaciones.
5. Muestra cuántas calificaciones se ingresaron y el promedio.

## 🧰 Conceptos aplicados
Listas · `append()` · ciclo `for` con `range()` · recorrer una lista con `for` · acumulador (`+=`) · `print()` con `end` · cálculo de promedio · `int()` y `float()` · f-strings

## 💻 Código

```python
# Promedio de calificaciones

# --- Datos iniciales ---
calificaciones = []          # lista vacía donde se guardan las calificaciones
suma_calificaciones = 0      # acumulador para calcular el promedio
cant_calificaciones = int(input('Indique cuantas calificaciones desea agregar a la lista: '))

# --- Registrar las calificaciones ---
# El ciclo se repite una vez por cada calificación; i+1 muestra el número desde 1
for i in range(cant_calificaciones):
    calificacion = float(input(f'{i+1} - Califiacion que desea agregar al lista: '))
    calificaciones.append(calificacion)      # agrega la calificación al final de la lista
    suma_calificaciones += calificacion      # la suma al acumulador

# --- Mostrar las calificaciones ingresadas ---
# end='  ' evita el salto de línea y deja dos espacios entre cada valor
for i in calificaciones:
    print(i, end = '  ')

# --- Calcular el promedio: suma total entre la cantidad de calificaciones ---
promedio_califiaciones = suma_calificaciones / cant_calificaciones

# --- Mostrar el resultado ---
print(f'\nUsted ingreso {cant_calificaciones} califiaciones.')
print(f'El promedio es: {promedio_califiaciones}')
```
