## 🎯 Planteamiento
Se necesita un sistema bancario básico que permita a un cliente entrar a su cuenta y realizar operaciones. El sistema debe confirmar si el usuario desea continuar, proteger el acceso con un PIN y ofrecer un menú para consultar el saldo, depositar y retirar dinero, validando en cada paso los datos que se ingresan.

## ⚙️ Qué hace el programa
1. Pregunta si el usuario desea continuar (Si/No) y repite la pregunta hasta recibir una respuesta válida.
2. Si responde "no", sale del sistema.
3. Si responde "si", pide el PIN. Tiene **3 intentos**; al fallar los tres, la cuenta se bloquea.
4. Con el PIN correcto, muestra un menú que se repite hasta que el usuario elija salir:
   - **Consultar saldo:** muestra el saldo disponible.
   - **Depositar:** suma el monto al saldo (debe ser mayor que 0).
   - **Retirar:** resta el monto del saldo, siempre que sea mayor que 0 y no supere el saldo disponible.
   - **Salir:** termina el programa.
5. Si el usuario elige una opción que no existe, le avisa y vuelve a mostrar el menú.

## 🧰 Conceptos aplicados
`input()` · `float()` · `lower()` y `strip()` · ciclos `while` (validación y menú) · operadores lógicos `and` / `not` · `if` / `elif` / `else` anidados · banderas (`True` / `False`) · contador de intentos · acumuladores (`+=` y `-=`) · f-strings con formato `:.2f`

## 💻 Código

```python
# Sistema bancario

# --- Datos de la cuenta ---
usuario = 'Jose'
pin_correcto = '1234'    # PIN de acceso a la cuenta
saldo = 5000.00          # saldo inicial en RD$

print('\n******SISTEMA BANCARIO******\n')

# --- Primera pregunta ---
# strip() quita espacios y lower() pasa a minúsculas, para que "SI " y "si" valgan igual
continuar_sistema = input('Desea continuar dentro del sistema bancario? Si/No.').lower().strip()

# --- Validación: repite la pregunta mientras la respuesta no sea "si" ni "no" ---
while continuar_sistema != 'si' and continuar_sistema != 'no':
    print('Ingrese una opcion valida Si/No.')
    continuar_sistema = input('Desea continuar dentro del sistema bancario? Si/No.').lower().strip()

# --- Decisión: salir o continuar ---
if continuar_sistema == 'no':
    print('Saliendo del sistema...')
else:
    print('Continuamos dentro del sistema...\n')

    # --- Acceso con PIN: máximo 3 intentos ---
    intentos = 0
    acceso = False
    while intentos < 3 and not acceso:
        pin = input('Ingrese su PIN: ').strip()
        if pin == pin_correcto:
            acceso = True                    # PIN correcto: se permite el acceso
        else:
            intentos += 1                    # suma un intento fallido
            print(f'PIN incorrecto. Intentos restantes: {3 - intentos}')

    # --- Si agotó los intentos, la cuenta se bloquea ---
    if not acceso:
        print('Cuenta bloqueada por demasiados intentos.')
    else:
        print(f'\nBienvenido, {usuario}')

        # --- Menú de operaciones: se repite hasta que el usuario elija salir ---
        en_sistema = True
        while en_sistema:
            print('\n--- MENU ---')
            print('1. Consultar saldo')
            print('2. Depositar')
            print('3. Retirar')
            print('4. Salir')
            opcion = input('Seleccione una opcion: ').strip()

            if opcion == '1':
                print(f'Saldo disponible: RD${saldo:.2f}')

            elif opcion == '2':
                monto = float(input('Monto a depositar RD$: '))
                if monto > 0:
                    saldo += monto           # suma el depósito al saldo
                    print(f'Deposito exitoso. Nuevo saldo: RD${saldo:.2f}')
                else:
                    print('El monto debe ser mayor que 0.')

            elif opcion == '3':
                monto = float(input('Monto a retirar RD$: '))
                if monto <= 0:
                    print('El monto debe ser mayor que 0.')
                elif monto > saldo:
                    print('Fondos insuficientes.')
                else:
                    saldo -= monto           # resta el retiro al saldo
                    print(f'Retiro exitoso. Nuevo saldo: RD${saldo:.2f}')

            elif opcion == '4':
                en_sistema = False           # termina el ciclo del menú
                print('Saliendo del sistema...')

            else:
                print('Opcion no valida. Elija 1, 2, 3 o 4.')
```
