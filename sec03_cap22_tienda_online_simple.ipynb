## 🎯 Planteamiento
Una tienda online necesita mostrar la ficha de un producto con su precio, cantidad en inventario y disponibilidad. Cuando el producto se agota, la ficha debe actualizarse para reflejar que ya no está disponible.

## ⚙️ Qué hace el programa
1. Guarda en variables los datos del producto: nombre, precio, stock y disponibilidad.
2. Muestra la ficha del producto con su inventario inicial.
3. Actualiza el stock a cero y marca el producto como no disponible.
4. Muestra la ficha actualizada.

## 🧰 Conceptos aplicados
Variables · tipos de datos (`str`, `float`, `int`, `bool`) · `print()` · reasignación de variables

## 💻 Código

```python
# Tienda online

# --- Datos iniciales del producto ---
nombre_producto = 'Leche'          # nombre del producto
precio_producto = 15.00            # precio de venta
cantidad_stock_producto = 5        # unidades en inventario
disponibilidad = True              # el producto está disponible

# --- Mostrar la ficha del producto ---
print("*****TIENDA ONLINE*****\n")
print('Nombre del producto:', nombre_producto)
print('Precio del producto:', precio_producto)
print('Cantidad disponible producto:', cantidad_stock_producto)
print('Disponibilidad:', disponibilidad)
print()                            # línea en blanco para separar las dos fichas

# --- El producto se agotó: actualizamos el inventario ---
cantidad_stock_producto = 0        # ya no quedan unidades
disponibilidad = False             # el producto deja de estar disponible

# --- Mostrar la ficha actualizada ---
print("*****TIENDA ONLINE*****\n")
print('Nombre del producto:', nombre_producto)
print('Precio del producto:', precio_producto)
print('Cantidad disponible producto:', cantidad_stock_producto)
print('Disponibilidad:', disponibilidad)
```
