# Sistema generador de ID único

📌 **Sección 5 · Capítulo 53**

## 🎯 Planteamiento
Se necesita un sistema que genere un identificador para cada persona a partir de su nombre, su apellido y su año de nacimiento. El ID debe tener un formato fijo y un componente aleatorio que lo haga distinto para cada usuario.

## ⚙️ Qué hace el programa
1. Pide al usuario el nombre, el apellido y el año de nacimiento (YYYY).
2. Toma las dos primeras letras del nombre y del apellido, en mayúsculas.
3. Toma los dos últimos dígitos del año de nacimiento.
4. Genera un número aleatorio de cuatro dígitos.
5. Une todo en un ID de 10 caracteres con el formato `NNAAYY####`, donde `NN` son las letras del nombre, `AA` las del apellido, `YY` el año y `####` el número aleatorio.

## 🧰 Conceptos aplicados
`input()` · `strip()` · `upper()` · slicing de cadenas (`[0:2]`) · módulo `random` con `randint()` · f-strings

## 💻 Código

```python
# Sistema de generador de ID único

# randint genera un número entero aleatorio dentro de un rango
from random import randint

# --- Entrada de datos por teclado ---
nombre = input('Ingrese su nombre:')
apellido = input('Ingrese su apellido:')
anio_nacimiento = input('Ingrese su año de nacimiento (YYYY):')

# --- Construir cada parte del ID ---
# Quita espacios, pasa a mayúsculas y toma las 2 primeras letras ([0:2])
inciales_nombre = nombre.strip().upper()[0:2]
inciales_apellido = apellido.strip().upper()[0:2]

# Toma los 2 últimos dígitos del año: de "1995" se queda con "95" ([2:4])
ultimos_anios = str(anio_nacimiento.strip()[2:4])

# Número aleatorio de 4 dígitos entre 1000 y 9999
codigo_unico = randint(1000, 9999)

# --- Unir todas las partes en el ID final ---
ID = f'{inciales_nombre}{inciales_apellido}{ultimos_anios}{codigo_unico}'

# --- Mostrar el resultado ---
print(ID)
```
