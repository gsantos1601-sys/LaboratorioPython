## Descripción
Simula la reserva de una habitación de hotel. Se guardan los datos del cliente y de la estancia en variables, se muestra la reserva con vista al mar y luego se modifican la tarifa y la vista para mostrar la misma reserva sin vista al mar.

## Temas
- Variables y tipos de datos (`str`, `int`, `float`, `bool`)
- Mostrar información con `print()`
- Reasignar el valor de una variable

```python
# Reserva de hoteles (Sec. 3, Cap. 20)

# --- Datos de la reserva con vista al mar ---
nombre_cliente = 'Diony'     # str: texto
dias_estancia = 5            # int: número entero
tarifa_diaria = 20.50        # float: número decimal
vista_al_mar = True          # bool: verdadero o falso

# --- Mostrar la reserva con vista al mar ---
print("RESERVA DE HOTELES CON VISTA AL MAR\n")  # \n deja una línea en blanco después del título
print('Nombre del cliente:', nombre_cliente)
print('Dias de estancia:', dias_estancia)
print('Tarifa diaria:', tarifa_diaria)
print('Incluye vista al mar?', vista_al_mar)

# --- Cambiamos los datos: ahora la habitación no tiene vista al mar ---
tarifa_diaria = 15           # reasignamos la tarifa (ahora es un entero)
vista_al_mar = False         # ya no incluye vista al mar

# --- Mostrar la reserva sin vista al mar ---
print()                      # línea en blanco para separar las dos reservas
print("RESERVA DE HOTELES SIN VISTA AL MAR\n")
print('Nombre del cliente:', nombre_cliente)
print('Dias de estancia:', dias_estancia)
print('Tarifa diaria:', tarifa_diaria)
print('Incluye vista al mar?', vista_al_mar)
print()                      # línea en blanco final
```
