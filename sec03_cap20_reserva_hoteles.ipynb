# Reserva de hoteles

📌 **Sección 3 · Capítulo 20**

## 🎯 Planteamiento
Un hotel necesita mostrar el resumen de la reserva de un cliente. La reserva inicial incluye vista al mar, pero el cliente decide cambiar a una habitación sin vista al mar con una tarifa más baja. El programa debe mostrar ambas versiones de la reserva.

## ⚙️ Qué hace el programa
1. Guarda en variables los datos de la reserva: cliente, días de estancia, tarifa diaria y vista al mar.
2. Muestra la reserva con vista al mar.
3. Actualiza la tarifa diaria y quita la vista al mar.
4. Muestra la reserva actualizada sin vista al mar.

## 🧰 Conceptos aplicados
Variables · tipos de datos (`str`, `int`, `float`, `bool`) · `print()` · reasignación de variables

## 💻 Código

```python
# Reserva de hoteles

# --- Datos de la reserva con vista al mar ---
nombre_cliente = 'Diony'     # nombre del cliente
dias_estancia = 5            # cantidad de días de la estancia
tarifa_diaria = 20.50        # tarifa por día
vista_al_mar = True          # la habitación incluye vista al mar

# --- Mostrar la reserva con vista al mar ---
print("RESERVA DE HOTELES CON VISTA AL MAR\n")
print('Nombre del cliente:', nombre_cliente)
print('Dias de estancia:', dias_estancia)
print('Tarifa diaria:', tarifa_diaria)
print('Incluye vista al mar?', vista_al_mar)

# --- El cliente cambia a una habitación sin vista al mar ---
tarifa_diaria = 15           # nueva tarifa, más baja
vista_al_mar = False         # ya no incluye vista al mar

# --- Mostrar la reserva sin vista al mar ---
print()                      # línea en blanco para separar las dos reservas
print("RESERVA DE HOTELES SIN VISTA AL MAR\n")
print('Nombre del cliente:', nombre_cliente)
print('Dias de estancia:', dias_estancia)
print('Tarifa diaria:', tarifa_diaria)
print('Incluye vista al mar?', vista_al_mar)
print()
```
