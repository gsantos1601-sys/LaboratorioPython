# Inventario de productos

📌 **Sección 10 · Capítulo 175**

## 🎯 Planteamiento
Se requiere un sistema interactivo por consola para administrar un inventario de productos. El programa debe permitir listar los productos almacenados, agregar nuevos registros garantizando IDs únicos y autoincrementables, buscar productos por su ID y validar las entradas numéricas del usuario para evitar errores en tiempo de ejecución.

## ⚙️ Qué hace el programa
1. Muestra un menú de opciones principal (mostrar, agregar, buscar y salir).
2. Valida mediante funciones dedicadas (`try-except`) que los datos ingresados (opciones, ID, stock, precio) sean numéricos y no negativos.
3. Asigna un ID automático al agregar un nuevo producto calculando el máximo ID actual más uno.
4. Muestra la lista completa de productos con su stock y precio en formato de pesos dominicanos (RD$).
5. Busca un producto específico por su ID e informa si no existe.
6. Permite al usuario decidir si desea volver al menú o finalizar la ejecución.

## 🧰 Conceptos aplicados
Funciones · Diccionarios · Listas de diccionarios · Manejo de excepciones (`try-except`) · Ciclo `while` e instrucciones de control (`break`, `continue`) · Expresión generadora con `max()` · Formateo de cadenas con f-strings · Métodos de cadenas (`strip()`, `capitalize()`, `lower()`)

## 💻 Código


```python
# Base de datos inicial almacenada como una lista de diccionarios
inventarios = [
    {'id': 1, 'Producto': 'Leche', 'Stock': 10, 'Precio': 50.99},
    {'id': 2, 'Producto': 'Pan', 'Stock': 5, 'Precio': 9.99},
    {'id': 3, 'Producto': 'Mantequilla', 'Stock': 3, 'Precio': 5.99}
]


def mostrar_menu():
    """Despliega las opciones principales del sistema en pantalla."""
    print('\n**** INVENTARIO DE PRODUCTOS ****')
    print('''
1. Mostrar inventario
2. Agregar nuevo producto
3. Buscar producto por ID
4. Salir
    ''')


def leer_entero(mensaje):
    """Solicita un valor al usuario y valida que sea un número entero válido."""
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print('Debe ingresar un número entero.')


def leer_decimal(mensaje):
    """Solicita un valor al usuario y valida que sea un número decimal/flotante válido."""
    while True:
        try:
            return float(input(mensaje))
        except ValueError:
            print('Debe ingresar un número válido.')


def volver_al_menu():
    """Pregunta al usuario si desea continuar navegando en el menú de opciones."""
    while True:
        respuesta = input('\nDesea volver al menu de opciones? Si/No: ').lower().strip()
        if respuesta in ('si', 'sí'):
            return True
        if respuesta == 'no':
            return False
        print('Responda únicamente Si o No.')


def siguiente_id():
    """Calcula automáticamente el siguiente ID disponible (máximo ID actual + 1)."""
    return max((p['id'] for p in inventarios), default=0) + 1


def imprimir_producto(p):
    """Formatea e imprime los datos individuales de un producto."""
    print(f'ID: {p["id"]} | Producto: {p["Producto"]} | '
          f'Cantidad disponible: {p["Stock"]} | Precio: RD$ {p["Precio"]}')


def mostrar_inventario():
    """Recorre y muestra todos los productos registrados en el inventario."""
    print('\n**** INVENTARIO ****')
    for producto in inventarios:
        imprimir_producto(producto)


def agregar_producto():
    """Solicita los datos necesarios y registra un nuevo producto en la lista."""
    print('\n*** AGREGAR PRODUCTO ***')
    nombre = input('Ingrese el producto: ').strip().capitalize()
    cantidad = leer_entero('Ingrese la cantidad disponible: ')
    precio = leer_decimal('Ingrese el precio: ')

    # Validación de montos no negativos
    if cantidad < 0 or precio < 0:
        print('La cantidad y el precio no pueden ser negativos.')
        return

    # Inserción del nuevo producto con ID autogenerado
    inventarios.append({
        'id': siguiente_id(),
        'Producto': nombre,
        'Stock': cantidad,
        'Precio': precio
    })
    print('Producto agregado correctamente.')


def buscar_producto():
    """Busca un producto dentro del inventario por su número de ID."""
    id_buscar = leer_entero('Introduce un ID: ')
    for producto in inventarios:
        if producto['id'] == id_buscar:
            imprimir_producto(producto)
            return
    print('Ese ID no existe.')


# ==========================================
# BUCLE PRINCIPAL DE EJECUCIÓN
# ==========================================
while True:
    mostrar_menu()
    opcion = leer_entero('Seleccione una opción: ')

    if opcion == 4:
        print('Saliendo del sistema...\n')
        break
    elif opcion == 1:
        mostrar_inventario()
    elif opcion == 2:
        agregar_producto()
    elif opcion == 3:
        buscar_producto()
    else:
        print('Digite una opción válida.')
        continue

    # Confirmación para permanecer en el menú o salir
    if not volver_al_menu():
        print('El sistema fue finalizado correctamente.')
        break
```

# ✅ Resultado
Ejemplo de ejecución con interacción del menú, registro de productos con ID automático, búsqueda por ID, manejo de entradas inválidas (textos y valores negativos) y salida controlada del sistema.
