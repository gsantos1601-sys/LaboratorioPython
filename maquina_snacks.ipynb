# Máquina de snacks

📌 **Sección 10 · Capítulo 179**

## 🎯 Planteamiento
Se requiere simular el funcionamiento de una máquina expendedora de snacks interactiva por consola. El programa debe permitir visualizar el catálogo de productos disponibles, realizar compras validando saldo y existencias en inventario, solicitar dinero incremental si el monto es insuficiente, descontar el stock actualizado y emitir una factura detallada con el total gastado, dinero ingresado y devuelta o saldo restante.

## ⚙️ Qué hace el programa
1. Presenta un menú de opciones principal para interactuar con la máquina expendedora.
2. Muestra los productos con su ID, cantidad en existencia y precio en formato de pesos dominicanos (RD$).
3. Valida la búsqueda de productos por ID verificando su existencia y disponibilidad de stock en el inventario.
4. Solicita el pago al usuario y acumula el saldo hasta cubrir el costo del producto deseado.
5. Procesa la compra reduciendo el stock en 1 unidad y guardando el artículo en el historial de compras.
6. Genera un desglose impreso de la factura con los productos adquiridos, el total pagado y el cambio final devuelto.

## 🧰 Conceptos aplicados
Funciones con parámetros y retornos · Listas de diccionarios · Métodos de diccionarios (`.get()`) · Manejo de excepciones (`try-except`) · Ciclos `while` e instrucciones de control (`break`) · Recursividad para reintentos de búsqueda · Formateo numérico y de moneda con f-strings · Estructura `if __name__ == '__main__':`

## 💻 Código

```python

# SIMULADOR DE MÁQUINA DE SNACKS
print('\n**** MAQUINA DE SNACKS ****')

# Base de datos inicial de productos disponibles almacenada como lista de diccionarios
productos = [
    {'id': 1, 'Producto': 'Papas', 'Stock': 0, 'Precio': 30.00},
    {'id': 2, 'Producto': 'Cocacola', 'Stock': 10, 'Precio': 25.00},
    {'id': 3, 'Producto': 'Agua', 'Stock': 10, 'Precio': 20.00},
    {'id': 4, 'Producto': 'Mani', 'Stock': 10, 'Precio': 10.00},
    {'id': 5, 'Producto': 'Gomitas', 'Stock': 10, 'Precio': 5.00}
]

# Variables globales para llevar el control del carrito de compras y del dinero ingresado
productos_comprados = []
saldo = 0.0
total_ingresado = 0.0


def mostrar_menu():
    """Despliega en consola las opciones disponibles del menú principal."""
    print('\n MENU:')
    print('''
1. Mostrar snacks
2. Comprar snack
3. Mostrar Factura
4. Salir
    ''')


def leer_entero(mensaje):
    """Solicita una entrada al usuario y valida que sea un número entero válido mediante try-except."""
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print('Debe ingresar un número entero.')


def mostrar_productos():
    """Recorre e imprime el catálogo completo de productos con su ID, inventario y precio."""
    print('\n******* SNACKS *******')
    for prod in productos:
        print(f'ID: {prod.get("id")} | Producto: {prod.get("Producto")} | '
              f'Cantidad disponible: {prod.get("Stock")} | Precio: RD$ {prod.get("Precio")}')


def buscar_producto():
    """
    Solicita un ID de producto, valida si existe en el catálogo y si tiene stock disponible.
    Aplica recursividad volviéndose a llamar si el producto no existe o está agotado.
    """
    id_buscar = leer_entero('Introduce un ID: ')
    for producto in productos:
        if producto.get('id') == id_buscar:
            if producto["Stock"] > 0:
                print('Producto existe y quedan disponibles')
                return producto
            else:
                print('Producto existe pero no hay disponible')
                return buscar_producto()
    print('No existe un producto con ese ID')
    return buscar_producto()


def solicitar_dinero(saldo):
    """Solicita y valida que el dinero ingresado sea mayor a cero, sumándolo al saldo acumulado."""
    while True:
        try:
            dinero_recibido = float(input('Ingrese el dinero para la compra: '))
            if dinero_recibido <= 0:
                print('Debe ingresar un valor mayor a 0')
            else:
                saldo += dinero_recibido
                return saldo, dinero_recibido
        except ValueError:
            print('Debe ingresar un valor numerico')


def comprar(saldo, producto):
    """Procesa la transacción: descuenta el precio del saldo, reduce el stock y guarda el producto comprado."""
    if saldo >= producto.get('Precio'):
        saldo -= producto.get('Precio')
        producto['Stock'] -= 1
        productos_comprados.append(producto)
        print('Compra exitosa.')
    else:
        faltante = producto.get('Precio') - saldo
        print(f'Saldo insuficiente. Te faltan RD$ {faltante:.2f}')
    return saldo


def factura(saldo, total_ingresado):
    """Imprime el desglose final de los productos adquiridos, el total, el pago realizado y el cambio devuelto."""
    if not productos_comprados:
        print('Aún no has comprado nada.')
        return
    print('\n******* FACTURA *******')
    total = 0
    for item in productos_comprados:
        print(f'{item.get("Producto")} ... RD$ {item.get("Precio"):.2f}')
        total += item.get('Precio')
    print(f'Total: RD$ {total:.2f}')
    print(f'Pagó con: RD$ {total_ingresado:.2f}')
    print(f'Cambio/Saldo restante: RD$ {saldo:.2f}')
    print(f'\nHistorial de compras: {productos_comprados}')


# ==========================================
# BUCLE PRINCIPAL DE EJECUCIÓN
# ==========================================
if __name__ == '__main__':

    while True:
        mostrar_menu()
        opcion = leer_entero('Seleccione una opción: ')

        if opcion == 4:
            print('Saliendo del sistema...\n')
            break
        elif opcion == 1:
            mostrar_productos()
        elif opcion == 2:
            producto = buscar_producto()
            # Solicita dinero repetidamente hasta cubrir el costo del producto seleccionado
            while saldo < producto.get('Precio'):
                saldo, dinero_recibido = solicitar_dinero(saldo)
                total_ingresado += dinero_recibido
            saldo = comprar(saldo, producto)
        elif opcion == 3:
            factura(saldo, total_ingresado)
        else:
            print('Digite una opción válida.')
```

## ✅ Resultado
Ejemplo de ejecución interactiva mostrando consulta del catálogo, intento de compra de producto agotado, ingreso de saldo insuficiente, actualización automática de inventarios, generación de factura con cambio e historial de ítems adquiridos.
