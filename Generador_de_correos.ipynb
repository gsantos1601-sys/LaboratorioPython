# Generador de correos

📌 **Sección 4 · Capítulo 41**

## 🎯 Planteamiento
Una empresa necesita crear correos corporativos a partir del nombre completo del empleado y del nombre de la compañía. Los datos llegan sin formato: con espacios de más y con mayúsculas. El programa debe devolver un correo con formato estándar.

## ⚙️ Qué hace el programa
1. Limpia los espacios sobrantes del nombre.
2. Cambia el espacio entre nombre y apellido por un punto.
3. Convierte el nombre a minúsculas.
4. Quita los espacios del nombre de la compañía y lo pasa a minúsculas.
5. Une todo con el formato `nombre.apellido@compañía.dominio`.

## 🧰 Conceptos aplicados
`strings` · `strip()` · `replace()` · `lower()` · encadenamiento de métodos · f-strings

## 💻 Código

```python
# Generador de correos

# --- Datos de entrada (llegan sin formato) ---
nombre = ' Gerardo Narvaez '   # espacios de más al inicio y al final
compania = 'La Sirena'         # nombre con espacio en medio
ext = '.com.do'                # extensión del correo

# --- Normalizar el nombre: limpiar, unir con punto y pasar a minúsculas ---
nombre = nombre.strip().replace(' ', '.').lower()

# --- Normalizar la compañía: quitar espacios y pasar a minúsculas ---
compania = compania.replace(' ', '').lower()

# --- Armar el correo con el formato final ---
correo_normalizado = f'{nombre}@{compania}{ext}'

# --- Mostrar el resultado ---
print(correo_normalizado)
```

## ✅ Resultado


gerardo.narvaez@lasirena.com.do
