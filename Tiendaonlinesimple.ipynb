## Descripción
Simula la ficha de un producto de una tienda online. Se guardan los datos del producto en variables, se muestran en pantalla y luego se actualiza el inventario para reflejar que el producto se agotó.

## Temas
- Variables y tipos de datos (`str`, `float`, `int`, `bool`)
- Mostrar información con `print()`
- Modificar el valor de una variable

```python
# --- Datos iniciales del producto ---
nombre_producto = 'Leche'          # str: texto
precio_producto = 15.00            # float: número decimal
cantidad_stock_producto = 5        # int: número entero
disponibilidad = True              # bool: verdadero o falso
```

```python
# --- Mostrar la ficha del producto ---
print("*****TIENDA ONLINE*****\n")  # \n deja una línea en blanco después del título
print('Nombre del producto:', nombre_producto)
print('Precio del producto:', precio_producto)
print('Cantidad disponible producto:', cantidad_stock_producto)
print('Disponibilidad:', disponibilidad)
print()  # línea en blanco para separar las dos fichas
```

```python
# --- Actualizamos el inventario: el producto se agotó ---
cantidad_stock_producto = 0        # reasignamos el valor de la variable
disponibilidad = False             # ya no está disponible
```

```python
# --- Mostrar la ficha actualizada ---
print("*****TIENDA ONLINE*****\n")
print('Nombre del producto:', nombre_producto)
print('Precio del producto:', precio_producto)
print('Cantidad disponible producto:', cantidad_stock_producto)
print('Disponibilidad:', disponibilidad)
```

## Explicación

**Variables:** cada una guarda un tipo de dato distinto: `'Leche'` es texto (`str`), `15.00` es decimal (`float`), `5` es entero (`int`) y `True` es booleano (`bool`).

**`print()`:** muestra texto y variables separados por coma. Todos los `print('Etiqueta:', variable)` siguen el mismo patrón; lo único que cambia es la etiqueta y la variable que se muestra.

**Actualización:** al asignar `cantidad_stock_producto = 0` y `disponibilidad = False`, las variables toman un valor nuevo. El nombre y el precio no cambian, por eso solo esos dos datos son distintos en la segunda ficha.
