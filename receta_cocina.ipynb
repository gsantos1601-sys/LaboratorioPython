## 🎯 Planteamiento
Se necesita un programa que registre una receta de cocina. El usuario escribe el nombre, los ingredientes y el nivel de dificultad, y el programa asigna automáticamente el tiempo de preparación según esa dificultad.

## ⚙️ Qué hace el programa
1. Pide al usuario el nombre de la receta, los ingredientes y la dificultad (Facil/Media/Alta).
2. Guarda los tiempos de preparación posibles en una lista: 10, 20 y 30 minutos.
3. Muestra el nombre, los ingredientes y la dificultad.
4. Elige el tiempo según la dificultad: Facil, 10 minutos; Media, 20 minutos; cualquier otra, 30 minutos.

## 🧰 Conceptos aplicados
`input()` · listas e índices · `if` / `elif` / `else` · operador de comparación `==` · f-strings

## 💻 Código

```python
# Reto: Receta de cocina (con condicionales)

# --- Entrada de datos por teclado ---
nombre_receta_cocina = input('Ingrese el nombre de la receta a preparar: ')
ingredientes_receta_cocina = input('Ingrese los ingredientes de la receta a preparar: ')
dificultad_preparacion = input('Indique la difucultad (Facil/Media/Alta)? ')

# --- Tiempos de preparación posibles, en minutos ---
# Posición 0 = Facil, posición 1 = Media, posición 2 = Alta
tiempo_prep_minutos = [10, 20, 30]

# --- Mostrar los datos de la receta ---
print('\n****RECETA DE COCINA****')
print(f'Receta: {nombre_receta_cocina}')
print(f'Ingredientes: {ingredientes_receta_cocina}')
print(f'Nivel de dificultad: {dificultad_preparacion}')

# --- Asignar el tiempo según la dificultad ---
if dificultad_preparacion == 'Facil':
    print(f'Tiempo de preparacion: {tiempo_prep_minutos[0]} minutos')   # 10 minutos
elif dificultad_preparacion == 'Media':
    print(f'Tiempo de preparacion: {tiempo_prep_minutos[1]} minutos')   # 20 minutos
else:
    print(f'Tiempo de preparacion: {tiempo_prep_minutos[2]} minutos')   # 30 minutos (Alta)
```
